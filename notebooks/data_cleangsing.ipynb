### proses pencarian outlier,missing value dan juga duplikasi data telah cukup sekarang kita masuk ke dalam pembersihan datanya

#### menghilangakn dupilkasi data karna murni 100% itu sudah sangat besar kemungkinan human eror jadi solusi kita akan melakukan syntax drop.dupilcate saja


```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

##### Sudah terlihat bahwa duplikasi data sudah 0

```python
df.columns = df.columns.str.lower()#mengubah header column jadi huruf kecil
for col in df.select_dtypes(include='object').columns:
  df[col]=df[col].str.lower()#untuk mempermudah kita rubah tulisan tipe data objek jadi huruf kecil semua menggunakan for looping
```

```python
df=df[(df['kuantitas']>0) & (df['harga_satuan_idr']>0)]
```

```python
df.describe()
```

##### oke tahap selanjutnya kita akan memperbaiki mis value nya

```python
# 2. Isi data kosong (Missing Value) agar data penjualannya tidak terbuang
df['region'] = df['region'].fillna('unknown')
df['sektor_industri'] = df['sektor_industri'].fillna('lainnya')
```

```python
df.info()
```

```python
df.isnull().sum()
```

##### dari dua info diatas telah di simpulkan bahwa missing value juga sudah kita perbaiki

#### sekarang kita akan coba mengubah tipe data yang tidak sesuai

```python
# 3. Ubah kolom tanggal dari teks (object) menjadi format waktu (datetime)
df['tanggal_transaksi'] = pd.to_datetime(df['tanggal_transaksi'])
```

#### oke semua otlier missing value begitu juga duplikasi data sudah kita selesaikan untuk saya biasanya langkah paling akhir yang selalu saya lakukan adalah pengecekan typo

```python
for col in ['region', 'sektor_industri', 'kategori_produk', 'status_transaksi']:
    print(df[col].value_counts())
    print("-" * 30)
```

##### bisa kita liat ada beberpa yang missing value seperti forklip ekskavator

```python
df['kategori_produk'] = df['kategori_produk'].replace({
    'ekskavator': 'excavator',
    'forklip': 'forklift'
})
```

```python
for col in ['region', 'sektor_industri', 'kategori_produk', 'status_transaksi']:
    print(df[col].value_counts())
    print("-" * 30)
```
